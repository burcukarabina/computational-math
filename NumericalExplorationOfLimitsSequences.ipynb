# Numerical Exploration of Limits Through Sequences

In this activity, we will explore what it means for a sequence to **converge**, to **diverge to infinity**, or to **fail to converge**.

The main idea is:

> We do not only care about the first few terms of a sequence.  
> We care about what happens in the **tail** of the sequence.

Let's recall the definition of a tail:

Let $\{a_n\}$ be a given sequence and $k\in\mathbb{N}$. 

The subsequence $\left\{a_{k}, a_{k+1}, a_{k+2}, \ldots\right\}$ is called the **tail** of $\left\{a_{n}\right\}$ with cutoff $k$, and is the sequence $\{a_n\}$ with the first $k-1$ points deleted.

In plain English, the **tail** of a sequence means the terms far out in the sequence, after many terms have already passed.

## A Sequence That Converges

Consider the sequence $\{a_n\}$ defined by

$$
a_n = 1 + \frac{1}{n}.
$$

Let's look at the first 10 terms.

```python
for n in range(1, 11):
    a = 1 + 1/n
    print("a_" + str(n) + " =", a)
```

As you can see, the value of the terms seems to be getting close to $1.1$.

Now let's look farther out in the sequence. These are terms from the **tail** of the sequence, terms from $a_{50}$ to $a_{60}$.

```python
for n in range(50, 61):
    a = 1 + 1/n
    print("a_" + str(n) + " =", a)
```

But wait, now, the value of the terms seems to be getting close to $1.01$. Maybe we should look further down the tail, terms from $a_{500000}$ to $a_{500010}$.

```python
for n in range(500000, 500010):
    a = 1 + 1/n
    print("a_" + str(n) + " =", a)
```

The terms get closer and closer to this $1$. In this case, we say

$$
\lim_{n\to\infty} \left(1+\frac{1}{n}\right)=1.
$$

## A Sequence Can Converge Even If It Oscillates

This activity addresses a common misconception that an oscillating sequence cannot converge.

Consider the sequence

$$
a_n = \frac{(-1)^n}{n}.
$$

This sequence changes sign: positive, negative, positive, negative, and so on.

Let's take a look at the first $10$ terms:

```python
for n in range(1, 10):
    a = ((-1)**n)/n
    print("a_" + str(n) + " =", a)
```

The sequence oscillates, but the terms are getting closer to 0.

Let's look at the tail.

```python
for n in range(50, 61):
    a = ((-1)**n)/n
    print("a_" + str(n) + " =", a)
```

The code below plots the end behavior of the sequence, confirming that it oscillates while converging to a limit.

```python
n_values = []
a_values = []

for n in range(1, 101):
    a = ((-1)**n)/n
    n_values.append(n)
    a_values.append(a)

plt.plot(n_values, a_values, marker='o')
plt.axhline(0)
plt.xlabel("n")
plt.ylabel("a_n")
plt.title("The sequence a_n = (-1)^n / n")
plt.show()
```

This sequence converges to 0:

$$
\lim_{n\to\infty}\frac{(-1)^n}{n}=0.
$$

The signs keep alternating, but the **tail** gets closer and closer to 0.

## Divergence to Infinity

Now consider the sequence

$$
a_n = 2n^3.
$$

Let's look at some terms.

```python
for n in range(1, 11):
    a = 2*n**3
    print("a_" + str(n) + " =", a)
```

Now let's look at a later part of the sequence.

```python
for n in range(100, 111):
    a = 2*n**3
    print("a_" + str(n) + " =", a)
```

It seems like the terms of the sequence keeps growing without a bound, and this sequence does **not** approach a finite number.

The terms keep getting larger and larger. We say the sequence **diverges to infinity**, and we write

$$
a_n \to \infty \quad \text{ or } \quad \lim_{n \to \infty} a_n = \infty
$$

This does **not** mean the sequence reaches infinity, because infinity is not a number. It means the terms become arbitrarily large.

## A Sequence That Does Not Converge

Now consider

$$
a_n = (-1)^n.
$$

Let's look at the first several terms.

```python
for n in range(1, 11):
    a = (-1)**n
    print("a_" + str(n) + " =", a)
```

This sequence keeps jumping between $-1$ and $1$.

Let's look at the tail.

```python
for n in range(50, 61):
    a = (-1)**n
    print("a_" + str(n) + " =", a)
```

Does the tail settle down near one number? Clearly not! For this reason, this sequence does not converge.

This sequence does **not** converge.

It also does **not** diverge to infinity.

Instead, it keeps oscillating between two values forever.
